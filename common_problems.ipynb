```python
import os
from PIL import Image
import torch
import matplotlib.pyplot as plt
```

```python
# visualize the tensors as images, 
# prompt_text,tensor_path_sd35,image_index_sd35
def load_images_from_examples(examples):
    for example in examples:
        prompt_text, tensor_path_sd35, image_index_sd35 = example.split(',')
        image_index_sd35 = int(image_index_sd35)
        data = torch.load(os.path.join('/mnt/lustre/work/hein/hof217/stable_diffusion/outputs', tensor_path_sd35))
        
        # Extract the image tensor from the dictionary
        tensor_sd35 = data['image']
        image_sd35 = Image.fromarray((tensor_sd35.permute(1, 2, 0).numpy() * 255).astype('uint8'))
        plt.figure()
        plt.title(prompt_text)
        plt.imshow(image_sd35)
        plt.axis('off')
        plt.show()
```
