```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
def show_all_images(output_dir):
    """
    Show all images one by one with their prompt_text and image_index_sd35.
    """
    csv_path = os.path.join(output_dir, "metadata.csv")
    df = pd.read_csv(csv_path)

    for _, row in df.iterrows():
        print(f"Index: {row['image_index_sd35']}")
        print(f"Prompt: {row['prompt_text']}")
        img = mpimg.imread(row['png_path_sd35'])
        plt.imshow(img)
        plt.axis("off")
        plt.show()
```

```python
import torch
import os
import matplotlib.pyplot as plt
import glob

def show_all_tensors(output_dir):
    """
    Show all images by loading directly from tensor files.
    Uses tensor data exclusively without relying on metadata.csv.
    
    Args:
        output_dir: Base directory containing tensor files
    """
    # Find all tensor files in the tensor directory
    tensor_dir = os.path.join(output_dir, "tensors", "sd35")
    tensor_files = sorted(glob.glob(os.path.join(tensor_dir, "*.pt")))
    
    if not tensor_files:
        print(f"No tensor files found in {tensor_dir}")
        return
    
    print(f"Found {len(tensor_files)} tensor files")
    
    for index, tensor_path in enumerate(tensor_files):
        print(f"\nImage {index}")
        
        # Load tensor data
        tensor_data = torch.load(tensor_path)
        
        # Extract image tensor and metadata
        if isinstance(tensor_data, dict):
            # If tensor_data is a dictionary with 'image' key
            if 'image' in tensor_data:
                img_tensor = tensor_data['image']
                
                # Print available metadata
                for key, value in tensor_data.items():
                    if key != 'image' and not torch.is_tensor(value):
                        print(f"{key}: {value}")
            else:
                # Try to find the first tensor in the dict
                tensor_keys = [k for k, v in tensor_data.items() if torch.is_tensor(v)]
                if tensor_keys:
                    img_tensor = tensor_data[tensor_keys[0]]
                    print(f"Using tensor from key: {tensor_keys[0]}")
                else:
                    print(f"No image tensor found in {tensor_path}")
                    continue
        elif torch.is_tensor(tensor_data):
            # If tensor_data is directly a tensor
            img_tensor = tensor_data
        else:
            print(f"Unexpected data type in {tensor_path}: {type(tensor_data)}")
            continue
        
        # Process the image tensor for display
        if len(img_tensor.shape) == 4:  # [batch, channels, height, width]
            img_tensor = img_tensor[0]  # Take first image if batched
            
        if len(img_tensor.shape) == 3:  # [channels, height, width]
            # Rearrange channels to match matplotlib's expected format
            img_tensor = img_tensor.permute(1, 2, 0)  # -> [height, width, channels]
        
        # Convert to numpy and normalize if needed
        img_np = img_tensor.cpu().numpy()
        
        # Normalize if needed
        if img_np.max() > 1.0:
            img_np = img_np / 255.0
        
        # Display the image
        plt.figure(figsize=(10, 10))
        plt.imshow(img_np)
        plt.axis("off")
        
        # Print tensor file name and shape
        print(f"Tensor file: {os.path.basename(tensor_path)}")
        print(f"Tensor shape: {img_tensor.shape}")
        
        plt.show()
```
