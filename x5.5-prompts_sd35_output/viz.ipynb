```python
def visualize_tensor_file(tensor_path):
    """
    Visualize all images in a tensor file along with their associated prompts.
    
    Args:
        tensor_path (str): Path to the tensor file (.pt)
    """
    if not os.path.exists(tensor_path):
        print(f"Tensor file not found: {tensor_path}")
        return
    
    try:
        # Load tensor data
        tensor_data = torch.load(tensor_path)
        
        # Extract images and prompts
        images = tensor_data.get('images', [])
        prompts = tensor_data.get('prompts', [])
        
        if not images:
            print(f"No images found in tensor file: {tensor_path}")
            return
            
        n_images = len(images)
        print(f"Found {n_images} images in {os.path.basename(tensor_path)}")
        
        # Display each image with its prompt
        for i, img_tensor in enumerate(images):
            fig, axes = plt.subplots(1, 2, figsize=(15, 5), gridspec_kw={'width_ratios': [1, 2]})
            
            # Display image
            axes[0].imshow(img_tensor.permute(1, 2, 0))  # Convert from [C,H,W] to [H,W,C]
            axes[0].set_title(f"Image {i+1}")
            axes[0].axis('off')
            
            # Display prompt if available
            axes[1].axis('off')
            prompt_text = prompts[i] if i < len(prompts) else "No prompt available"
            axes[1].text(0, 0.5, f"Prompt: {prompt_text}", fontsize=12, verticalalignment='center')
            
            plt.tight_layout()
            plt.show()
            
    except Exception as e:
        print(f"Error loading tensor file: {e}")
```

```python
# code to only show the first image+metadata from the tensor file
def visualize_first_image(tensor_path):
    """
    Visualize only the first image in a tensor file along with its associated prompt.
    
    Args:
        tensor_path (str): Path to the tensor file (.pt)
    """
    if not os.path.exists(tensor_path):
        print(f"Tensor file not found: {tensor_path}")
        return
    
    try:
        # Load tensor data
        tensor_data = torch.load(tensor_path)
        
        # Extract images and prompts
        images = tensor_data.get('images', [])
        prompts = tensor_data.get('prompts', [])
        
        if not images:
            print(f"No images found in tensor file: {tensor_path}")
            return
        
        # Get only the first image and prompt
        img_tensor = images[0]
        prompt_text = prompts[0] if prompts else "No prompt available"
        
        # Display the image with its prompt
        fig, axes = plt.subplots(1, 2, figsize=(15, 5), gridspec_kw={'width_ratios': [1, 2]})
        
        # Display image
        axes[0].imshow(img_tensor.permute(1, 2, 0))  # Convert from [C,H,W] to [H,W,C]
        axes[0].set_title(f"First image in {os.path.basename(tensor_path)}")
        axes[0].axis('off')
        
        # Display prompt
        axes[1].axis('off')
        axes[1].text(0, 0.5, f"Prompt: {prompt_text}", fontsize=12, verticalalignment='center')
        
        plt.tight_layout()
        plt.show()
        
        # Return the image and prompt in case needed for further processing
        return img_tensor, prompt_text
            
    except Exception as e:
        print(f"Error loading tensor file: {e}")
        return None, None
```
