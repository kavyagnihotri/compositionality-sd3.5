```python
from datasets import load_dataset
ds = load_dataset("eugene6923/Self-Bench")
ds = ds["train"]
print(len(ds))
```

```python
position_ds = ds.filter(lambda x: x["tag"] == "position")
```

```python
# fiiter such that the prompts are unique
def get_unique_samples(ds):
    seen_prompts = set()
    unique_samples = []
    for sample in ds:
        prompt = sample["prompt"]
        if prompt not in seen_prompts:
            seen_prompts.add(prompt)
            unique_samples.append(sample)
    return unique_samples

unique_position_ds = get_unique_samples(position_ds)
print(f"Number of unique position samples: {len(unique_position_ds)}")
```

```python
# print the max and min length of the prompts with the prompt itself
max_length = 0
min_length = float('inf')
max_prompt = ""
min_prompt = ""
for sample in unique_position_ds:
    prompt_length = len(sample["prompt"])
    if prompt_length > max_length:
        max_length = prompt_length
        max_prompt = sample["prompt"]
    if prompt_length < min_length:
        min_length = prompt_length
        min_prompt = sample["prompt"]

print(f"Max prompt length: {max_length}")
print(f"Max prompt: {max_prompt}")  
```
